# VisualQA India — Kaggle runner

Before running: choose a GPU accelerator and enable Internet in Kaggle's Notebook options. Replace `YOUR_USERNAME` below after uploading this project to GitHub.

```python
import os, subprocess, sys

GITHUB_REPO = "https://github.com/YOUR_USERNAME/visualqa-india.git"
PROJECT_DIR = "/kaggle/working/visualqa-india"

if "YOUR_USERNAME" in GITHUB_REPO:
    raise ValueError("Replace YOUR_USERNAME with your GitHub username first.")
if not os.path.exists(PROJECT_DIR):
    subprocess.run(["git", "clone", "--depth", "1", GITHUB_REPO, PROJECT_DIR], check=True)
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", f"{PROJECT_DIR}/requirements.txt"], check=True)
os.chdir(PROJECT_DIR)
print("Project ready:", os.getcwd())
```

```python
import torch
assert torch.cuda.is_available(), "Enable a GPU accelerator in Kaggle Notebook options."
print(torch.cuda.get_device_name(0))
print(f"CUDA memory: {torch.cuda.get_device_properties(0).total_memory / 2**30:.1f} GiB")
```
